# Blue Team Metrics Analysis — Seminario 11

Este notebook analiza resultados **exportados** del laboratorio defensivo. No genera tráfico de ataque ni recibe objetivos externos.


```python
from google.colab import files
import json, pandas as pd, matplotlib.pyplot as plt

uploaded = files.upload()
print('Archivos:', list(uploaded))
```

```python
def load_result(filename):
    raw = uploaded[filename].decode('utf-8')
    return json.loads(raw)

before_name = next(name for name in uploaded if 'before' in name.lower())
after_name = next(name for name in uploaded if 'after' in name.lower())
before = load_result(before_name)
after = load_result(after_name)
before, after
```

```python
def row(label, data):
    return {
        'phase': label,
        'requests': data.get('requests', 0),
        'rps': data.get('requests_per_second', 0),
        'mean_ms': data.get('latency_ms', {}).get('mean', 0),
        'p95_ms': data.get('latency_ms', {}).get('p95', 0),
        'max_ms': data.get('latency_ms', {}).get('max', 0),
        'status_counts': data.get('status_counts', {})
    }

df = pd.DataFrame([row('before', before), row('after', after)])
df
```

```python
ax = df.set_index('phase')[['rps','p95_ms','max_ms']].plot(kind='bar', figsize=(10,5))
ax.set_title('Comparación antes vs después')
ax.set_ylabel('Valor')
plt.xticks(rotation=0)
plt.show()
```
